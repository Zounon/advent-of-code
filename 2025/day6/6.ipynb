```python
input = []
with open('input') as file:
# with open('sample') as file:
    while line := file.readline():
        input.append(line.replace('\n','').split())

nums = [[int(x) for x in line] for line in input[:-1]]
ops = input[-1]

t = [[None for _ in range(len(nums))] for _ in range(len(nums[0]))]

for row_idx, row in enumerate(nums):
    for col_idx, col in enumerate(row):
        t[col_idx][row_idx] = col

s=0
for idx,l in enumerate(t):
    if ops[idx] == '+':
        s+=reduce(lambda x,y: x+y, l) 
    else:
        s+=reduce(lambda x,y: x*y, l) 
        
print(f"Day 6 Part 1:  {s}")
```
