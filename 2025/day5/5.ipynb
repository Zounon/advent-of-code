```python
from itertools import *
from functools import *
ranges = []
ids = []
flag = True
# with open('sample') as file:
with open('input') as file:
    while line := file.readline():
        if line == '\n':
            flag = False
            continue
        if flag:
            ranges.append(list(map(int,line.replace('\n','').split('-'))))
        else:
            ids.append(int(line))


ranges.sort()
rs = [range(l,r+1) for (l,r) in ranges]
```

```python
ranges = sorted(ranges)
s = [ranges[0]]

idx = 1
while idx < len(ranges):
    last = s.pop()
    last_l, last_r = last

    curr = ranges[idx]
    curr_l, curr_r = curr

    if last_r >= curr_l:
        s.append([min(curr_l,last_l), max(last_r,curr_r)])

    else:
        s.append(last)
        s.append(curr)
    idx += 1

x = 0
for l,r in s:
    x += r-l+1
print(x)
```
