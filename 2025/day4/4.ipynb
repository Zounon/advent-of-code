```python

from itertools import *
from typing import TypeAlias

Cell: TypeAlias = tuple[int, int]

def is_paper_cell(grid: list[list[chr]], cell: Cell) -> bool:
    return grid[cell[0]][cell[1]] == '@'

def valid_neighbours(cell: Cell, max_rows: int, max_cols: int) -> list[Cell]:
    is_in_bounds = lambda c: 0 <= c[0] < max_rows and 0 <= c[1] < max_cols
    is_not_origin = lambda c: c != cell
    return list(filter(
        lambda c: is_in_bounds(c) and is_not_origin(c),
        [(cell[0] + row_offset, cell[1] + col_offset) for (row_offset, col_offset) in product(range(-1,2), range(-1,2))]
        ))

def forklift_accessable_cell(grid: list[list[chr]], cell: Cell) -> bool:
    max_rows = len(grid)
    max_cols = len(grid[0])
    forklift_limit = 4
    return forklift_limit > sum([is_paper_cell(grid, neighbour) for neighbour in valid_neighbours(cell,max_rows,max_cols)])

def next_state(grid: list[list[chr]]) -> list[list[chr]]:
    max_rows = len(grid)
    max_cols = len(grid[0])

    def next_cell(cell :Cell) -> str:
        if forklift_accessable_cell(grid, cell) and is_paper_cell(grid,cell):
            return 'x'
        return grid[cell[0]][cell[1]]
    
    return [''.join(next_cell((r,c)) for c in range(max_cols)) for r in range(max_rows)]

def final_state(grid: list[list[chr]]) -> list[list[chr]]:
    next_grid = next_state(grid)
    if next_grid == grid:
        return grid
    return final_state(next_grid)

def num_of_x(grid: list[list[chr]]) -> int:
    max_rows = len(grid)
    max_cols = len(grid[0])
    return sum([grid[r][c] == 'x' for (r,c) in product(range(max_rows), range(max_cols))])


print(num_of_x(final_state(og_input)))
```
