```python
from itertools import *

points = list(map(eval, open('input')))

# add first point at the end to wrap around
pairs = list(pairwise(points + [points[0]])) 
max_area = 0

for (xa,ya), (xb,yb) in combinations(points, 2):
    if xa > xb:
        xa, xb = xb, xa

    if ya > yb:
        ya, yb = yb, ya
    
    area = (xb-xa+1) * (yb-ya+1)

    if area > max_area:
        for (xc,yc), (xd,yd) in pairs:
            if xc > xd:
                xc, xd = xd, xc

            if yc > yd:
                yc, yd = yd, yc
            
            if xc < xb and yc < yb and xd > xa and yd > ya:
                break
        else:
            max_area = area

print(max_area)
```

```python
def area(a,b):
    x1,y1 = a
    x2,y2 = b
    # return sqrt(pow((x1-x2),2) * pow((y1-y2),2))
    height = abs(y1-y2)+1
    length = abs(x1-x2)+1
    return height * length

# area(2,5,9,7)
# area(2,5,11,1)
area((2,5),(11,1))
```
