```python

from itertools import *
```

```python
shapes = []
with open('inputshapes') as file:
    while line := file.readline().strip():
        print(line)
        line = line.split(',')
        shape = []
        for l in line:
            shape.append([int(x) for x in l])
        shapes.append(shape)
# shapes[0]
```

```python
shape_sizes = []
for shape in shapes:
    shape_sizes.append(sum(list(chain.from_iterable(shape))))
shape_sizes
```

```python
areas = []
shape_reqs = []
with open('input') as file:
    while line := file.readline().strip():
        # print(line)
        area, reqs = line.split(":")
        areas.append([int(x) for x in area.split('x')])
        shape_reqs.append([int(x) for x in reqs.split()])
print(len(areas), len(shape_reqs))
```

```python
def trivially_unsolveable(area, req):
    area_size = area[0] * area[1]
    req_size = sum(starmap(int.__mul__, zip(shape_sizes, req)))
    return area_size < req_size

trivially_unsolveable(areas[0], shape_reqs[0])
```

```python
t = [trivially_unsolveable(area, req) for area, req in zip(areas, shape_reqs)]
sum(t)
```

```python
problems = [(area, req) for area, req in zip(areas, shape_reqs) if not trivially_unsolveable(area,req)]
problems.sort()
len(problems)
```
