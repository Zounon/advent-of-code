```python
input = []
with open('sample') as file:
# with open('input') as file:
    while line := file.readline():
        a,b = line.strip().split(':')
        input.append((int(a),tuple(map(int, b.split(' ')[1:]))))

input[0]
```

```python
def f(start, vals, og_val):
    if start == og_val and not vals:
        return True
    
    if start > og_val:
        return False
    
    if not vals:
        return False
    
    return f(start * vals[0], vals[1:], og_val) or f(start + vals[0], vals[1:], og_val) 
```

```python
def f2(start, vals, og_val):
    if start == og_val and not vals:
        return True
    
    if start > og_val:
        return False
    
    if not vals:
        return False
    
    return f2(start * vals[0], vals[1:], og_val) or \
            f2(start + vals[0], vals[1:], og_val) or \
            f2(int(str(start) + str(vals[0])), vals[1:], og_val) 
```

```python
cnt = 0
cnt2 = 0
for start, vals in input:
    # print(start, vals)
    if f(0,vals, start):
        cnt += start
        cnt2 += start
    elif f2(0, vals, start):
        cnt2 += start
    
print(cnt)
print(cnt2)
```
