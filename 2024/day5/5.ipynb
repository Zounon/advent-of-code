```python
# input = open('sample').read().split('\n\n')
input = open('input').read().split('\n\n')
rules = [tuple(map(int, line.split('|'))) for line in input[0].split('\n')]
updates = [[int(x) for x in line.split(',')] for line in input[1].split()]
# print(rules)
# print(updates)
```

```python
M = {}
for rule in rules:
    a,b = rule
    if a in M:
        M[a].append(b)
    else:
        M[a] = [b]
```

```python
from itertools import pairwise
from functools import cmp_to_key
```

```python
bad_updates = []
good_updates = []
for update in updates:
    update_is_good = True
    if all([end in M.get(start, []) for start, end in pairwise(update)]):
        good_updates.append(update)
    else:
        bad_updates.append(update)
part1 = 0
for u in good_updates:
    part1 += u[len(u)//2]

def page_cmp(a,b):
    return -1 if (b in M.get(a, [])) else 1

part2 = 0
for update in bad_updates:
    fixed = sorted(update, key=cmp_to_key(page_cmp))
    part2 += fixed[len(fixed)//2]

print(f"Part1: {part1}\nPart2: {part2}")
```
