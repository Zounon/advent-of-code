```python
input = []
file_name = 'sample'
file_name = 'input'
with open(file_name) as file:
    while line := file.readline():
        input.append([c for c in line.strip()])

m =  {r+c*1j: char for r,row in enumerate(open(file_name)) for c,char in enumerate(row.strip())}
start_pos = [pos for pos in m if m[pos] == '^'][0]


num_rows = len(input)
num_cols = len(input[0])
og_guard_pos = [(row, col) for row in range(num_rows) for col in range(num_cols) if input[row][col] == '^'][0]

# print(num_rows, num_cols)
# print(guard_pos)
# m
```

```python
guard_pos = og_guard_pos
visited = set([guard_pos])

def in_bounds(row, col):
    return (0 <= row < num_rows and 0 <= col < num_cols)

dir = [(-1,0),(0,1),(1,0),(0,-1)]
dchar = ['^', '>', 'v', '<']
didx = 0

# while in_bounds(guard_pos):
while True:
    visited.add(guard_pos)
    x,y = guard_pos
    a,b = dir[didx]
    new_pos = (x+a, y+b)
    # print(guard_pos, dir[didx])
    if not in_bounds(*new_pos):
        break
    
    if input[x+a][y+b] == '#':
        didx += 1
        didx %= 4
    else:
        guard_pos = new_pos
    
print(len(visited))
```

```python

def in_bounds_c(pos):
    return 0 <= pos.real < num_rows and 0 <= pos.imag < num_cols

# Learned from 4HbQ's solution
# https://www.reddit.com/r/adventofcode/comments/1h7tovg/2024_day_6_solutions/m0o44m5/ 
def walk(m):
    pos = start_pos
    dir = -1 + 0j
    visited = set()
    while in_bounds_c(pos) and (pos,dir) not in visited:
    # while pos in m and (pos,dir) not in visited:
        visited.add((pos,dir))
        if m.get(pos+dir) == '#':
            dir *= -1j
        else:
            pos += dir
    return {pos for pos,_ in visited}, (pos,dir) in visited

path = walk(m)[0]
print(f"Part1: {len(path)}\nPart2: {sum(walk(m | {pos: '#'})[1] for pos in path)}")
```
