```python

input = []
with open('input') as file:
    while line := file.readline():
        input.append(list(line.replace('\n','')))

start_pos = input[0].index('S')
input[1][start_pos] = '|'

split_count = 0

row_idx = 2
while row_idx < len(input):
    last_row = input[row_idx-1]

    for col_idx, val in enumerate(input[row_idx]):
        if val == '.' and last_row[col_idx] == '|':
            input[row_idx][col_idx] = '|'
        elif val == '^' and last_row[col_idx] == '|':
            
            split_count +=1

            if col_idx-1 >= 0:
                input[row_idx][col_idx-1] = '|'
            if col_idx+1 < len(input[0]):
                input[row_idx][col_idx+1] = '|'
    
    row_idx += 1







print(split_count)
input
```
