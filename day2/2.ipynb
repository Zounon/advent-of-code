```python
def is_invalid_twice(x :int) -> bool:
    x = str(x)
    if len(x) % 2 != 0:
        return False
    
    half = len(x) // 2
    # print(x[:half] == x[half:])
    return x[:half] == x[half:]
```

```python
all_invalid = []

ranges = input.split(',')
for r in ranges:
    # print(r)
    [a,b] = r.split('-')
    for i in range(int(a),int(b)):
        # print(i)
        if is_invalid_twice(i):
            all_invalid.append(i)

# all_invalid
sum(all_invalid)
```

```python
def is_invalid(total_str :int) -> bool:
    total_str = str(total_str)
    # print(total_str)
    # total_str = total_str + total_str
    # print(total_str)

    sub_str = []
    for char in total_str:
        if char not in sub_str:
            sub_str.append(char)
        else:
            break
    
    if len(sub_str) == len(total_str):
        # all chars are unique -> no repetition
        return False
    
    if len(total_str) / len(sub_str) == len(total_str) // len(sub_str):
        # posible repetition
        for idx in range(len(total_str)):
            if total_str[idx] != sub_str[idx % len(sub_str)]:
                return False
        
        return True
    else:
        return False
```

```python
def is_invalid2(total_str: int) -> bool:
    total_str = str(total_str)

    sub_str_len = [ssl for ssl in range(1,len(total_str)) if len(total_str)/ssl == len(total_str)//ssl]

    for ssl in sub_str_len:
        mul = len(total_str) // ssl
        # print(ssl, mul)
        if total_str == (total_str[:ssl] * mul):
            return True
    return False
```

```python
all_invalid = []

ranges = input.split(',')
for r in ranges:
    # print(r)
    [a,b] = r.split('-')
    for i in range(int(a),int(b)):
        # print(i)
        if is_invalid2(i):
            all_invalid.append(i)

# all_invalid
sum(all_invalid)
```
