```python
# with open('sample') as file:
with open('input') as file:
    input = file.read()
input = input.split('\n')
```

```python
s=[]
for line in input:
    max_val = -1
    max_idx = -1
    for idx, val in enumerate(reversed(line)):
        val = int(val)
        if val >= max_val:
            max_val = val
            max_idx = len(line) - idx - 1
    # print()
    # print(max_val, max_idx)

    if max_idx == 0:
        left_max = max_val
        right_max = max(line[1:])
    elif max_idx == (len(line)-1):
        left_max = max(line[:-1])
        right_max = max_val
    else:
        left_max = max_val
        right_max = max(line[max_idx+1:])

    # print(line)
    # print(left_max,  right_max)
    s.append(int(left_max) * 10 + int(right_max))
print(sum(s))
```

```python
s = []
ss = []
for line in input:
    res = list(line[-12:])
    for new_val in reversed(line[:-12]):
        if new_val >= res[0]:
            res.insert(0, new_val)
            idx = 2
            while idx < len(res)+1:
                if (idx == len(res)):
                    res = res[:-1]
                    break
                elif (res[idx-1] < res[idx]):
                    res = res[:idx-1] + res[idx:]
                    break
                idx +=1

    v = 0
    for idx,val in enumerate(reversed(res)):
        v += int(val) * 10 ** idx
    ss.append(v)
print(sum(ss))
```
